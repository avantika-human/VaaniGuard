```python
# Cell 4
# Import Python's regular-expression module for pattern matching.
import re


# Define a function to extract monetary amounts from transcribed text.
# This function only extracts amounts as evidence.
# It does not independently decide whether the conversation is risky.
def extract_money_amounts(text):

    # Define patterns for common currency formats.
    money_patterns = [

        # Dollar amounts such as $1,450 or $50.
        r'\$\s?\d+(?:,\d{3})*(?:\.\d+)?',

        # Indian rupee symbol amounts such as ₹50,000.
        r'₹\s?\d+(?:,\d{3})*(?:\.\d+)?',

        # Amounts followed by currency names.
        r'\b\d+(?:,\d{3})*(?:\.\d+)?\s?(?:dollars?|rupees?|rs)\b',

        # Amounts written with thousand, lakh, million, or crore terms.
        r'\b\d+(?:\.\d+)?\s?(?:thousand|lakh|million|crore)\b'
    ]

    # Store all detected monetary amounts.
    detected_amounts = []

    # Search the text using each pattern.
    for pattern in money_patterns:

        # Find all matches while ignoring letter case.
        matches = re.findall(pattern, text.lower())

        # Add the matches to the result list.
        detected_amounts.extend(matches)

    # Remove duplicate amounts while preserving their original order.
    detected_amounts = list(dict.fromkeys(detected_amounts))

    # Return the extracted amounts as supporting evidence.
    return detected_amounts
```

```python
# Cell 5
# Define a function to combine and display all text-analysis results.
def display_analysis_result(transcribed_text, risk_indicators, money_amounts):

    # Combine all analysis outputs into one dictionary.
    analysis_result = {
        "transcription": transcribed_text,
        "risk_indicators": risk_indicators,
        "monetary_amounts": money_amounts
    }

    # Display the complete analysis result.
    print("Audio Analysis Report")
    print("---------------------")

    print("Transcription:")
    print(analysis_result["transcription"])

    print("\nRisk indicators:")
    print(analysis_result["risk_indicators"])

    print("\nMonetary amounts:")
    print(analysis_result["monetary_amounts"])

    # Return the complete result so it can be used later.
    return analysis_result
```

```python
# Cell 6
# Define a function to calculate a basic conversation-risk score.
# This score is based on suspicious language in the transcription.
# It is separate from AASIST's fake-probability score.

def calculate_risk_score(risk_indicators, money_amounts):

    # Assign weights to different suspicious conversation indicators.
    indicator_weights = {

        # Financial transaction indicators.
        "wire transfer": 3,
        "bank transfer": 3,
        "transfer money": 3,
        "send money": 3,
        "urgent payment": 3,
        "immediate payment": 3,

        # Sensitive information requests.
        "otp": 5,
        "verification code": 5,
        "share your pin": 5,
        "password": 5,
        "cvv": 5,

        # Account threats.
        "account blocked": 4,
        "account suspended": 4,
        "legal action": 4,

        # Remote-access requests.
        "remote access": 5,
        "screen sharing": 5,
        "anydesk": 5,
        "teamviewer": 5
    }

    # Start the conversation-risk score at zero.
    score = 0

    # Add the weight for every detected risk indicator.
    for indicator in risk_indicators:
        score += indicator_weights.get(indicator, 1)

    # Add a small supporting contribution when a monetary amount is present.
    if len(money_amounts) > 0:
        score += 2

    # Return the conversation-risk score.
    return score
```

```python
#cell 7
# Define a function to convert a numerical risk score into a risk category.
def classify_risk_level(score):

    # Very high scores indicate a critical-risk conversation.
    if score >= 10:
        return "CRITICAL"

    # Scores from 6 to 9 indicate high risk.
    elif score >= 6:
        return "HIGH"

    # Scores from 3 to 5 indicate moderate risk.
    elif score >= 3:
        return "MEDIUM"

    # Scores below 3 indicate low risk.
    else:
        return "LOW"
```

```python
#cell 8
# Define one function that performs the complete text-analysis pipeline.
def analyze_transcription(text):

    # Extract suspicious phrases from the transcription.
    indicators = extract_risk_indicators(text)

    # Extract monetary amounts from the transcription.
    amounts = extract_money_amounts(text)

    # Calculate the numerical risk score.
    score = calculate_risk_score(indicators, amounts)

    # Convert the numerical score into a risk category.
    level = classify_risk_level(score)

    # Return all results in one structured dictionary.
    return {
        "transcription": text,
        "risk_indicators": indicators,
        "monetary_amounts": amounts,
        "risk_score": score,
        "risk_level": level
    }
```

```python
# Define a general category-based analyzer.
def analyze_risk_categories(text):

    # Convert text to lowercase.
    text_lower = text.lower()

    # Store detected categories and matching phrases.
    detected_categories = {}
    total_score = 0

    # Check each category.
    for category, phrases in risk_categories.items():

        # Find all matching phrases in this category.
        matches = [
            phrase for phrase in phrases
            if phrase in text_lower
        ]

        # Special handling for financial requests.
        if category == "financial_request" and matches:

            # Financial language alone is not necessarily suspicious.
            suspicious_context = any(
                phrase in text_lower
                for phrase in (
                    risk_categories["urgency_pressure"]
                    + risk_categories["threat_intimidation"]
                    + risk_categories["credential_request"]
                    + risk_categories["remote_access"]
                    + risk_categories["secrecy_control"]
                )
            )

            # Ignore ordinary financial statements.
            if not suspicious_context:
                continue

        # Special handling for impersonation.
        if category == "impersonation" and matches:

            # Authority/institution mention alone is not necessarily suspicious.
            suspicious_context = any(
                phrase in text_lower
                for phrase in (
                    risk_categories["credential_request"]
                    + risk_categories["financial_request"]
                    + risk_categories["urgency_pressure"]
                    + risk_categories["threat_intimidation"]
                    + risk_categories["remote_access"]
                    + risk_categories["secrecy_control"]
                )
            )

            # Ignore harmless mentions of institutions or departments.
            if not suspicious_context:
                continue

        # Store only categories that have valid matches.
        if matches:
            detected_categories[category] = matches

            # Add the category weight only once.
            total_score += category_weights[category]

    # Return the category analysis.
    return {
        "categories": detected_categories,
        "category_score": total_score
    }
```

```python
#cell 11
#Combined indicators- Your account will be blocked. Share the OTP and transfer the money immediately."
#Base score = 10
# Contextual bonus = 4
# Final score = 14
#Why not just use the normal score?
#Because the normal score may treat categories independently:
# Define a function to calculate additional risk based on category combinations.

def calculate_contextual_bonus(detected_categories):
    # Extract only the detected category names.
    categories = set(detected_categories.keys())

    # Start with no contextual bonus.
    bonus = 0

    # Financial request combined with urgency is more suspicious.
    if "financial_request" in categories and "urgency_pressure" in categories:
        bonus += 3

    # Financial request combined with credential request is highly suspicious.
    if "financial_request" in categories and "credential_request" in categories:
        bonus += 4

    # Impersonation combined with threats is suspicious.
    if "impersonation" in categories and "threat_intimidation" in categories:
        bonus += 3

    # Remote access combined with financial request is highly risky.
    if "remote_access" in categories and "financial_request" in categories:
        bonus += 4

    # Credential requests combined with urgency are suspicious.
    if "credential_request" in categories and "urgency_pressure" in categories:
        bonus += 3

    # Secrecy combined with any other risk category increases suspicion.
    if "secrecy_control" in categories and len(categories) > 1:
        bonus += 2

    return bonus
```

```python
#cell 12 - calculates the conversation- risk score
#Example - Suppose the conversation contains:
# Base category score       = 10
# Contextual bonus          = 7
# Money bonus               = 2
# --------------------------------
# Final conversation score  = 19


"Your account will be blocked. Share the OTP and transfer ₹5000 immediately."
# Define a function to calculate the final risk score.
def calculate_final_risk_score(category_result, money_amounts):
    # Get the base score from detected risk categories.
    base_score = category_result["category_score"]

    # Get the detected categories.
    detected_categories = category_result["categories"]

    # Calculate the bonus for suspicious category combinations.
    contextual_bonus = calculate_contextual_bonus(detected_categories)

    # Add an additional score if a monetary amount is mentioned.
    money_bonus = 2 if len(money_amounts) > 0 else 0

    # Calculate the final score.
    final_score = base_score + contextual_bonus + money_bonus

    return {
        "base_category_score": base_score,
        "contextual_bonus": contextual_bonus,
        "money_bonus": money_bonus,
        "final_risk_score": final_score
    }
```

```python
#cell 12 - This is actually a good integration function. It brings all the previous text-analysis
# functions together into one pipeline.
# Define one complete function for analyzing any transcription.
# This produces the conversation-analysis report.
# It is separate from AASIST's fake-probability score.

def complete_text_analysis(text):
    # Detect specific suspicious phrases.
    indicators = extract_risk_indicators(text)

    # Detect monetary amounts.
    amounts = extract_money_amounts(text)

    # Detect broader risk categories.
    categories_result = analyze_risk_categories(text)

    # Calculate contextual combination bonus.
    contextual_bonus = calculate_contextual_bonus(
        categories_result["categories"]
    )

    # Calculate the final risk score.
    final_score_result = calculate_final_risk_score(
        categories_result,
        amounts
    )

    # Extract the final numerical score.
    final_score = final_score_result["final_risk_score"]

    # Classify the final score into a risk level.
    risk_level = classify_risk_level(final_score)

    # Return the complete analysis report.
    return {
        "transcription": text,
        "specific_indicators": indicators,
        "risk_categories": categories_result["categories"],
        "monetary_amounts": amounts,
        "base_category_score": categories_result["category_score"],
        "contextual_bonus": contextual_bonus,
        "money_bonus": final_score_result["money_bonus"],
        "final_risk_score": final_score,
        "final_risk_level": risk_level
    }
```

```python
# Cell 13

# ============================================================
# Detect specific intentions in the conversation
# ============================================================

# Risk categories describe WHAT type of suspicious language appeared.
# Intents describe WHAT the caller appears to be trying to make
# the listener do.

intent_phrases = {

    # Attempts to make the listener send or move money.
    "requesting_money": [
        "send money",
        "transfer money",
        "pay now",
        "make a payment",
        "complete payment",
        "transfer amount",
        "send amount",
        "approve transaction",
        "move money"
    ],

    # Attempts to obtain sensitive credentials or verification details.
    "requesting_credentials": [
        "tell me your otp",
        "share otp",
        "provide password",
        "tell me your pin",
        "share pin",
        "read the code",
        "give verification code",
        "share the code"
    ],

    # Language designed to pressure the listener into acting quickly.
    "creating_urgency": [
        "do immediately",
        "right now",
        "act immediately",
        "must do now",
        "before it is too late",
        "within ten minutes",
        "within 10 minutes"
    ],

    # Threats or consequences used to pressure the listener.
    "threatening": [
        "account blocked",
        "account will be blocked",
        "you will be arrested",
        "legal action",
        "police complaint",
        "account suspended"
    ],

    # Attempts to obtain access to the listener's device or banking app.
    "requesting_remote_access": [
        "download this app",
        "install this application",
        "share your screen",
        "remote access",
        "open your banking app"
    ]
}


# Define a function to detect the caller's apparent intentions.
def detect_intents(text):

    # Convert text to lowercase for case-insensitive matching.
    text_lower = text.lower()

    # Store detected intents and their supporting phrases.
    detected_intents = {}

    # Check every intent category.
    for intent, phrases in intent_phrases.items():

        # Find phrases belonging to the current intent.
        matched_phrases = [
            phrase for phrase in phrases
            if phrase in text_lower
        ]

        # Store only intents that have at least one match.
        if matched_phrases:
            detected_intents[intent] = matched_phrases

    # Return the detected intentions.
    return detected_intents
```

```python
#cell 14

#"Share your OTP and transfer money immediately."
# Detected intents:
# requesting_credentials
# requesting_money
# creating_urgency

# ============================================================
# STEP 2: Score the detected intentions
# ============================================================

intent_weights = {
    "requesting_money": 3,
    "requesting_credentials": 5,
    "creating_urgency": 2,
    "threatening": 4,
    "requesting_remote_access": 5
}


def calculate_intent_score(detected_intents):
    """Calculate a score based on the caller's likely intention."""
    score = 0

    for intent in detected_intents:
        score += intent_weights.get(intent, 0)

    return score


def calculate_intent_context_bonus(detected_intents):
    """Add extra points when multiple dangerous intentions occur together."""
    intents = set(detected_intents.keys())
    bonus = 0

    if "requesting_money" in intents and "creating_urgency" in intents:
        bonus += 3

    if "requesting_money" in intents and "requesting_credentials" in intents:
        bonus += 4

    if "requesting_remote_access" in intents and "requesting_money" in intents:
        bonus += 4

    if "requesting_credentials" in intents and "creating_urgency" in intents:
        bonus += 3

    return bonus
```

```python
#cell 15
# Transcription
#    ├── Specific indicators
#    ├── Monetary amounts
#    ├── Risk categories + category bonus
#    └── Intents + intent bonus
#              ↓
#        Final conversation-risk score
#              ↓
#         Risk level
# ============================================================
# STEP 3: Improve the final score using categories + intents
# ============================================================

def improved_text_analysis(text):
    """
    Run the complete text-analysis pipeline.

    This combines:
    1. Specific suspicious phrases
    2. Broad risk categories
    3. Monetary amounts
    4. Caller intentions
    5. Contextual combination bonuses
    """

    indicators = extract_risk_indicators(text)
    amounts = extract_money_amounts(text)
    category_result = analyze_risk_categories(text)
    intents = detect_intents(text)

    category_score = category_result["category_score"]
    category_bonus = calculate_contextual_bonus(
        category_result["categories"]
    )

    intent_score_value = calculate_intent_score(intents)
    intent_bonus = calculate_intent_context_bonus(intents)

    money_bonus = 2 if amounts else 0

    final_score = (
        category_score
        + category_bonus
        + intent_score_value
        + intent_bonus
        + money_bonus
    )

    final_level = classify_risk_level(final_score)

    return {
        "transcription": text,
        "specific_indicators": indicators,
        "risk_categories": category_result["categories"],
        "detected_intents": intents,
        "monetary_amounts": amounts,
        "category_score": category_score,
        "category_context_bonus": category_bonus,
        "intent_score": intent_score_value,
        "intent_context_bonus": intent_bonus,
        "money_bonus": money_bonus,
        "final_risk_score": final_score,
        "final_risk_level": final_level
    }
```

```python
#cell 16
# “I completed the bank transfer yesterday.” → likely normal
# “Transfer the money immediately and share your OTP.” → much more suspicious
#The cell tries to reduce the score when the speaker is merely describing a completed or past event.
# ============================================================
# STEP 4: Avoid false positives from normal conversations
# ============================================================

# These phrases often indicate that the speaker is describing
# a completed, failed, or historical event rather than requesting
# the listener to perform a risky action.

neutral_context_phrases = [
    "already completed",
    "completed yesterday",
    "completed last week",
    "i received",
    "i paid",
    "i transferred",
    "transaction completed",
    "transaction successful",
    "transfer failed",
    "payment failed",
    "was debited",
    "transaction history",
    "checking my balance"
]


# Detect phrases suggesting a neutral or past-event context.
def detect_neutral_context(text):

    # Convert text to lowercase for matching.
    text_lower = text.lower()

    # Return all neutral-context phrases found in the text.
    return [
        phrase for phrase in neutral_context_phrases
        if phrase in text_lower
    ]


# Calculate a small score adjustment for neutral conversations.
def calculate_neutral_adjustment(text, detected_intents):

    # Detect neutral or historical context phrases.
    neutral_phrases = detect_neutral_context(text)

    # Reduce the score only when:
    # 1. Neutral context is present.
    # 2. No actionable suspicious intention is detected.
    if neutral_phrases and not detected_intents:
        return -2

    # Otherwise, do not change the score.
    return 0
```

```python
# Cell 17/18
# Explanation dictionaries convert detected evidence into readable reasons.

category_explanations = {
    "financial_request":
        "The speaker appears to request a money transfer or payment.",

    "credential_request":
        "The speaker appears to request sensitive credentials or verification information.",

    "urgency_pressure":
        "The speaker is creating time pressure to force a quick decision.",

    "threat_intimidation":
        "The speaker uses threats or consequences to pressure the listener.",

    "remote_access":
        "The speaker requests remote access, screen sharing, or installation of an application.",

    "impersonation":
        "The speaker appears to impersonate an institution or trusted authority.",

    "secrecy_control":
        "The speaker attempts to prevent the listener from consulting others."
}


intent_explanations = {
    "requesting_money":
        "The apparent objective is to obtain money or authorize a financial transaction.",

    "requesting_credentials":
        "The apparent objective is to obtain OTPs, PINs, passwords, or other sensitive information.",

    "creating_urgency":
        "The speaker attempts to make the listener act immediately.",

    "threatening":
        "The speaker uses possible account, legal, or police consequences as pressure.",

    "requesting_remote_access":
        "The speaker attempts to gain access to the listener's device or banking application."
}


# Create readable explanations using only detected evidence.
def generate_risk_explanation(report):

    # Store explanation sentences.
    explanations = []

    # Explain detected risk categories.
    for category in report["risk_categories"]:

        if category in category_explanations:
            explanations.append(category_explanations[category])

    # Explain detected caller intentions.
    for intent in report["detected_intents"]:

        if intent in intent_explanations:
            explanations.append(intent_explanations[intent])

    # Mention monetary evidence if an amount was detected.
    if report["monetary_amounts"]:

        explanations.append(
            "A monetary amount was mentioned: "
            + ", ".join(report["monetary_amounts"])
        )

    # Provide a neutral explanation if no suspicious evidence was found.
    if not explanations:

        explanations.append(
            "No strong suspicious evidence was detected in the transcription."
        )

    # Return the explanation sentences.
    return explanations
```

```python
#cell 19
# ============================================================
# STEP 7: Decide what the system should do next
# ============================================================

# Convert the conversation-risk level into a practical system action.
def decide_system_action(risk_level):
    """
    LOW      -> no immediate action
    MEDIUM   -> continue monitoring and collect more context
    HIGH     -> trigger a warning
    CRITICAL -> trigger an urgent warning
    """

    if risk_level == "CRITICAL":
        return "TRIGGER_URGENT_WARNING"

    elif risk_level == "HIGH":
        return "TRIGGER_WARNING"

    elif risk_level == "MEDIUM":
        return "MONITOR_AND_COLLECT_MORE_CONTEXT"

    else:
        return "NO_ACTION"
```

```python
# Test a legitimate remote-support conversation.
sample_text = (
    "The university IT administrator scheduled a remote support session "
    "through the official helpdesk to fix my laptop."
)

report = improved_text_analysis(sample_text)
explanations = generate_risk_explanation(report)

print("Final Analysis Report")
print("---------------------")

for key, value in report.items():
    print(f"{key}: {value}")

print("\nWhy was this considered risky?")

for explanation in explanations:
    print("-", explanation)

print("\nRecommended system action:")
print(decide_system_action(report["final_risk_level"]))
```
